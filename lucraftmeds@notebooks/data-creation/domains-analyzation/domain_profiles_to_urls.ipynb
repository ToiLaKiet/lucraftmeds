# Convert domain profiles to domain URL lists

Đọc `domain_profiles.json`, gom các `url_pattern` của từng domain, loại pattern trùng và xuất dữ liệu dạng:

```json
[{"domain": "example.com", "urls": ["/article/*/*.html"]}]
```

````python
import json
from pathlib import Path

INPUT_PATH = Path("domain_profiling_output/domain_profiles.json")
OUTPUT_PATH = Path("domain_profiling_output/domain_urls.json")

if not INPUT_PATH.exists():
    raise FileNotFoundError(
        f"Không tìm thấy {INPUT_PATH.resolve()}. "
        "Hãy chạy notebook từ thư mục domains-analyzation."
    )

with INPUT_PATH.open(encoding="utf-8") as file:
    profiles = json.load(file)

len(profiles.get("domains", []))
````

````python
domain_urls = []

for domain_profile in profiles.get("domains", []):
    # dict giữ nguyên thứ tự và đồng thời loại URL pattern trùng.
    unique_patterns = {}

    for template in domain_profile.get("templates", []):
        url_pattern = template.get("url_pattern")
        if url_pattern:
            unique_patterns[url_pattern] = None

    domain_urls.append({
        "domain": domain_profile.get("domain", ""),
        "urls": list(unique_patterns),
    })

print(f"Số domain: {len(domain_urls):,}")
print(f"Tổng URL pattern: {sum(len(item['urls']) for item in domain_urls):,}")
domain_urls[:3]
````

````python
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

with OUTPUT_PATH.open("w", encoding="utf-8") as file:
    json.dump(domain_urls, file, ensure_ascii=False, indent=2)

print(f"Đã ghi file: {OUTPUT_PATH.resolve()}")
````
